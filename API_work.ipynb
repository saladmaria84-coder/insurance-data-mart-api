```python
import pandas as pd
import requests

path = r'C:\Users\Марьям\Downloads\\'

print('Загружаем файлы...')
df_contracts = pd.read_excel(path + 'CASE_CONTRACTS.xlsx')
df_losses = pd.read_excel(path + 'CASE_LOSSES.xlsx')
df_clients = pd.read_excel(path + 'experiment_group.xlsx')
print('Файлы успешно загружены!')


url = 'https://www.cbr-xml-daily.ru/daily_json.js'
response = requests.get(url)
data_currency = response.json()
usd_rate = data_currency['Valute']['USD']['Value']
print(f'Текущий курс USD: {usd_rate}')


df_mart = df_contracts.merge(
    df_clients, on='client_id', how='left'
).merge(df_losses, on='client_id', how='left')


if 'insurance_amount' in df_mart.columns:
  df_mart['insurance_amount_usd'] = df_mart['insurance_amount'] / usd_rate

if 'loss_payout_amt' in df_mart.columns:
  df_mart['loss_payout_usd'] = df_mart['loss_payout_amt'] / usd_rate

output_path = path + 'insurance_datamart_final.xlsx'
df_mart.to_excel(output_path, index=False)

print(f'Готово! Витрина сохранена в файл: {output_path}')
```
